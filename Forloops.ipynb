```python
# Multiplication TAble upto 10 values
num=int(input("Enter value"))
for i in range(1,11):
  print(f"{num}* {i}={num*i}")
```

```python
#Sum of digits in a given number
num=int(input("Enter number:"))
sum=0
while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10 # floor value

print("Sum of given number ", sum)
```

```python
#Reverse a number
num=int(input("Enter number:"))
sum=0
while(num!=0):
  rem=num%10
  sum=sum*10+rem
  num=num//10
print(sum)
```

```python
#Prime number or not
num=int(input("Enter a number:"))
if(num<=1):
  print("not a prime")
else:
  for i in range(2,int(num**0.5)+1):
    if(num%i==0):
      print("not a prime")
      break
  else:
      print("prime")
```

```python
#Program that prints prime number between given range
low=int(input("Enter lower range"))
high=int(input("Enter higher range"))
for i in range(low,high+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
    print(i)
```
