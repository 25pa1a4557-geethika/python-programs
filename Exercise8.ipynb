```python
#1.Program to find no of duplicates in list
l=list(map(int,input("Enter values:").split()))
s=set(l)
print(s)
print(len(l)-len(s))
```

```python
#2.Program to find no of elements having duplicate values by printing those elements
l2=[]
count=0
for i in l:
  if l.count(i)>1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
print(len(s))
```

```python
#Write a python program that prints missing numbers btw 1 to n
n=int(input("Enter n value:"))
l=list(map(int,input("Enter values:").split()))
print("missing numbers")
for i in range(1,n+1):
  if i not in l:
    print(i,end=" ")
```
