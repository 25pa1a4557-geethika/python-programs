Dictionaries in Python

```python
 #Dictionary Comprehension
d={i:i*i for i in range(1,6)}
print(d)
```

```python
#User defined dictionary
k = [i for i in input().split()]
v = [int(i) for i in input().split()]
d = {i: j for i, j in zip(k, v)}
print(d)
```

```python
#Dictionary Bulit in Functions
d={"sno":101,"name": "Sita", "a":10,"b":20,"c":30}
print("Dictionary ",d)
print("Length:",len(d))
print("Sorted Keys:",sorted(d))
print("Max:",max(d))
print("min:",min(d))
print("list keys",list(d))

x=dict(a=10,b=20)
print("new dict",x)
print("Any key exists:",any(d))
print("all ",all(d))
print("String ",str(d))
print(type(d))
```

```python
#Dictionary Methods
d={"rno":1 , "name":"john", "age":19,"Course":"ai&ds"}
print("name:",d.get("name"))
print("Keys:",d.keys())
print("Values:",d.values())
print("Items:",d.items())
d.update({"course":"ai&ml"})
print("After update:",d)
print("removed",d.pop("age"))
print("after pop ",d)
print("Removed pair:",d.popitem())
print("After popitem :",d)
d.setdefault("year","2nd year")
print("after setdefault:",d)
```
