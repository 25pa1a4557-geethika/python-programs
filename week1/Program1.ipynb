Operators in Python programms:

```python
#Arithmetic operations

x=int(input("enter value :"))
y=int(input("enter value :"))
print("Addition of two values is ",x+y)
print("subtraction of two values is",x-y)
print("Multiplication of two values is",x*y)
print("Division of two values is",x/y)
print(x%y)
print(x**y)
print(x//y)
```

```python
#Comparison operators

p=int(input("enter value:"))
q=int(input("enter value:"))
print(p==q)
print(p!=q)
print(p>q)
print(p<q)
print(p>=q)
print(p<=q)
```

```python
#Assignment operators
x = 10
x += 5
print(x)   # 15
x -= 3
print(x)   # 12
x *= 2
print(x)   # 24
x /= 4
print(x)   # 6.0
```

```python
#Logical operators

a=int(input("Enter a num:"))
b=int(input("Enter a num:"))
print(a and b)
print(a or b)

print(not(a > 3 and a < 10))
```

```python
#Identity operators
x = ["apple", "banana"]
y = [" grapes","orange"]
print(x is y)
print(x is not y)
```

```python
# Membership operators
fruits =["apple" , "banana","cherry"]
print("banana" in fruits)
print("papaya "not in fruits)
```

```python
# Bitwise operators
x=int(input("Enter value:"))
y=int(input("Enter value:"))

print(x&y)
print(x|y)
print(x^y)
print(~x)
print(x>>2)
print(x<<2)
```

```python
#Ternary Operator

score = int(input("Enter score:"))
print("Passed" if score >= 50 else "Failed")
```
