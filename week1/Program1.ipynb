Operators in Python programms:

```python
#Arithmetic operations

x=int(input("enter value :"))
y=int(input("enter value :"))
print("Addition of two values is ",x+y)
print("subtraction of two values is",x-y)
print("Multiplication of two values is",x*y)
print("Division of two values is",x/y)
print(x%y)
print(x**y)
print(x//y)
```

```python
#Comparison operators

p=int(input("enter value:"))
q=int(input("enter value:"))
print(p==q)
print(p!=q)
print(p>q)
print(p<q)
print(p>=q)
print(p<=q)
```

```python
#Logical operators

a=int(input("Enter a num:"))
b=int(input("Enter a num:"))
print(a and b)
print(a or b)
print(a or b)
```
