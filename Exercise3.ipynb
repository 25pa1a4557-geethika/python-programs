```python
#List Operations
l1=[1,2,3,"hello"]
l2=[6,7,8,"hai"]
print("concatenation :",l1+l2)
print("repitation:",l2*4)
print("indexing:",l2[3])
print("indexing:",l1[-1])
print("slicing:",l1[1:])
if 6 in l2:
  print("Found")
else:
  print("Not found")
print("Iteration:")
for i in l2:
     print(i)
```

```python
#List Bulit in Functions
l3=[11,22,33,44]
l4=[10,20,30]
print("Sum of l3 and l4:",sum(l4))
print("Maximum:",max(l3))
print("Minimum:",min(l4))
print("length of l3:",len(l3))
print("reverse of l4:",list(reversed(l4)))
print("sorted l3:",sorted(l3))
```

```python
#List Methods
l1.append("python")
print("After append l1 is :",l1)
l3.extend(l4)
print("After extend l3:",l3)
l2.insert(0,"java")
print("After insert l2:",l2)
print("Count method:",l3.count(30))
l3.pop(1)
print("Pop method on l3:",l3)
l3.insert(5,200)
l3.insert(0,11)
l3.remove(11)
print("Remove method on l3:",l3)
l3.sort()
print("Sort:",l3)
l2.reverse()
print("Reverse l2:",l2)
```

```python
#Program that prints sum of list elements without using sum
l=[10,20,30,25]
sum=0
for i in l:
  sum=sum+i

print("Sum of elements in list:",sum)
```

```python
#Print the list of elements at odd index
l=[10,20,30,40,50,60,70,80]
print("Odd index elements are:",l[1::2])
```

```python
#Program that prints even number
l = [10, 15, 20, 25, 30, 35]
print("Even numbers are:")
for i in l:
    if i % 2 == 0:
        print(i)
```
