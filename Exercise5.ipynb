```python

#Program that reverse a string using function
def reverse(s):
   return s[::-1]
s=input("Enter a string:")
print("Reverse string:",reverse(s))
```

```python
#palindrome or not using functions
def is_palindrome(num):
    original = num
    reverse = 0

    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num = num // 10

    return original == reverse


n = int(input("Enter a number: "))

if is_palindrome(n):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Program that reverse
def rev_str(s):
  x=""
  for i in range(len(s)-1,-1,-1):
    x=x+s[i]
  return x
print(rev_str("hello"))
```

```python
#Function that returns multiple values
def multiple_function(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_function(1,2,3,4,5,6,7)
print(f"the sum of numbers is {res_sum},the count is {res_count},the average is{res_avg:.2f}")
```
