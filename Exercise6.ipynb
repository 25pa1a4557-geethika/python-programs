                            TUPLES

```python
#Operations in Tuples
t1=(11,22,33,"python")
t2=(44,55,66,"java")
print("concatenation",t1+t2)
print("repetition",t1*3)
print("indexing",t1[3])
print("slicing",t2[1:4])
print("membership",20 in t1)
print("iteration")
for i in t2:
  print(i)
```

```python
#Methods in Tuples
t3=(1,2,3,4,5,6,7)
print("Count",t3.count(5))
print("Index of 4:",t3.index(4))
```

```python
#Bulit in functions in tuples
t4=(1,5,7,4,2,9,8)
print("Length of t1:",len(t1))
print("Maximum of t3:",max(t3))
print("Minimum of t4:",min(t4))
print("Sum of t3",sum(t3))
print("Sorted order of t4 ",sorted(t4))
```

```python
#Using Zip() function
name=("Ram","John","Sita")
marks=(85,90,94)
print(tuple(zip(name,marks)))
```

```python
#Write a program to count the number of vowels in a string (No control flow allowed)
s1=input("Enter a string:")
s2=s1.lower()
vowel_count=s2.count("a")+s2.count("e")+s2.count("i")+s2.count("o")+s2.count("u")
print(f"The number of vowels in given string is {vowel_count}")
```
