```python
#Write program that counts no of vowels
a=input("Enter:")
vowels="aeiou"
count=0
for char in a:
  if char.lower() in vowels:
    count=count+1
print(count)
```

```python
#program that prints reverse of number using function
def reverse(n):
  r=0
  while n>0:
    digit=n%10
    r=r*10+digit
    n=n//10
  return r
num=int(input("Enter number:"))
result=reverse(num)
print("Reverse :",result)
```
