```python
#Operations in set
s1={1,2,3,4,7}
s2={5,6,7,8,3}
print(s1 | s2)
print(s1 & s2)
print(s1 - s2)
print(s1 ^ s2)
```

```python
  # Methods in set
s3 = {1, 2, 3, 5, 6}
s4 = {2, 4, 6, 8, 10}
s5 = {1, 2, 3}
s3.add(7)
print("After add:", s3)
s4.update([9, 11])
print("Update:", s4)
s3.remove(3)
print("After remove():", s3)
s3.discard(9)
print("After discard:", s3)
x = s3.pop()
print("Popped element:", x)
print("After pop():", s3)
s5.clear()
print("After clear():", s5)
print("Union:", s3.union(s4))
print("Intersection:", s3.intersection(s4))
print("Difference:", s3.difference(s4))
print("Symmetric Difference:", s3.symmetric_difference(s4))
```
