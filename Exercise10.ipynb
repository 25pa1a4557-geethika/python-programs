```python
#Write a program that prints frequency of each element in a list in following
# format [1,2,3,4,1,5,6,1,7,2]   {1:3,2:2,3:1,4:1,5:1,6:1,7:1}
l=[1,2,3,4,1,5,6,1,7,2]
f={}
for ele in l:
  if ele in f:
    f[ele]+=1
  else:
    f[ele]=1
print(f)
```
