```python
#Uswing write mode
file=open("student.txt","w")
file.write("This is a new student file")
print("Student file is created")
file.close()
```

```python
#using read mode
with open("student.txt","r") as f:
  data=f.read()
  print(data)
```

```python
#Using append mode
with open("student.txt","a+") as f:
    f.writelines(['\napple', '\nGeethika', '\nAI&DS'])
    f.seek(0)
    print(f.readline())
    print(f.readlines())
    print(f.tell())
```

```python
#Write a program that copies content of one file to another file
with open("f1.txt","w") as f:
  f.writelines(["Hello World\n","This is source file\n","This content will be copied\n"])

with open("f1.txt","r") as source:
  data=source.read()

with open("f2.txt","w") as dest:
  dest.write(data)

print("File copied")
```

```python
#Write  program that prints each line of a file in reverse order
with open("f1.txt","r") as f1:
  for i in f1:
    print(i[::-1])
```

```python
#write to program to sort words in a file and put them in another file
with open("f1.txt","r") as f:
  data=f.read().split()
  sort_data = sorted(data, key=str.lower)
with open("f4.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
  print("done")

```

```python
#rb,wb binary file handling
import pickle
with open("s1.dat","wb") as f:
  data=[1,2,3,"hello"]
  pickle.dump(data,f)

with open("s1.dat","rb") as f2:
  d=pickle.load(f2)
  print(d)
```
